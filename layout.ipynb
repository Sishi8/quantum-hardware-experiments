```python
import qiskit_metal as metal
from qiskit_metal import designs, MetalGUI
from qiskit_metal.qlibrary.qubits.transmon_pocket import TransmonPocket
from qiskit_metal.qlibrary.sample_shapes.rectangle import Rectangle

# =====================================================
# 1. DESIGN (IMPORTANT FIX)
# =====================================================
design = designs.DesignPlanar(enable_renderers=False)
gui = MetalGUI(design)

# =====================================================
# 2. QUBIT CREATION FUNCTION
# =====================================================
def make_qubit(name, x, y):
    TransmonPocket(
        design,
        name,
        options=dict(
            pos_x=f'{x}mm',
            pos_y=f'{y}mm',
            pad_width='400um',
            pad_height='250um'
        )
    )

# =====================================================
# 3. HEAVY-HEX GEOMETRY (7 QUBITS)
# =====================================================
dx = 2.0
dy = 1.5

coords = {
    'Q0': (0, dy),
    'Q1': (-dx, 0),
    'Q2': (dx, 0),
    'Q3': (-dx, -dy),
    'Q4': (0, -2*dy),
    'Q5': (dx, -dy),
    'Q6': (0, -3*dy),
}

for name, (x, y) in coords.items():
    make_qubit(name, x, y)

# =====================================================
# 4. SIMPLE MANUAL BUS (NO ROUTER = NO ERRORS)
# =====================================================
def draw_bus(name, x, y, w, h):
    Rectangle(
        design,
        name,
        options=dict(
            pos_x=f'{x}mm',
            pos_y=f'{y}mm',
            width=f'{w}mm',
            height=f'{h}mm'
        )
    )

# =====================================================
# 5. CONNECTIONS (MANUAL CPW BUSES)
# =====================================================

# Top connections
draw_bus('b0', -1.0, 0.75, 1.5, 0.1)   # Q0 → Q1
draw_bus('b1',  1.0, 0.75, 1.5, 0.1)   # Q0 → Q2

# Vertical connections
draw_bus('b2', -2.0, -0.75, 0.1, 1.5)  # Q1 → Q3
draw_bus('b3',  2.0, -0.75, 0.1, 1.5)  # Q2 → Q5

# Middle connections
draw_bus('b4', -1.0, -1.5, 1.5, 0.1)   # Q3 → Q4
draw_bus('b5',  1.0, -1.5, 1.5, 0.1)   # Q5 → Q4

# Bottom connection
draw_bus('b6', 0.0, -3.0, 0.1, 1.5)    # Q4 → Q6

# =====================================================
# 6. BUILD + DISPLAY
# =====================================================
gui.rebuild()
gui.autoscale()
gui.show()
```
