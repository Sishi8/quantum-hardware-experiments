```python
import qiskit_metal as metal
from qiskit_metal import designs, MetalGUI
from qiskit_metal.qlibrary.qubits.transmon_pocket import TransmonPocket

# ------------------------------------------------
# DESIGN
# ------------------------------------------------
design = designs.DesignPlanar(enable_renderers=False)
gui = MetalGUI(design)

# ------------------------------------------------
# QUBITS ONLY (NO ROUTING)
# ------------------------------------------------
def make_qubit(name, x, y):
    return TransmonPocket(
        design,
        name,
        options=dict(
            pos_x=f'{x}mm',
            pos_y=f'{y}mm',
            pad_width='400um',
            pad_height='250um'
        )
    )

coords = {
    'Q0': (0, 2),

    'Q1': (-2, 1),
    'Q2': (2, 1),

    'Q3': (-2, -1),
    'Q4': (0, -1),
    'Q5': (2, -1),

    'Q6': (0, -3),
}

for name, (x, y) in coords.items():
    make_qubit(name, x, y)

# ------------------------------------------------
# BUILD
# ------------------------------------------------
gui.rebuild()
gui.autoscale()
```
